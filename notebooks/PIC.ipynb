# Feriados

```python
import requests
import pandas as pd

url = "https://date.nager.at/api/v3/PublicHolidays/2023/BR"

response = requests.get(url, timeout=30)
response.raise_for_status()  # gera erro se a chamada falhar

dados = response.json()
df_feriados = pd.DataFrame(dados)

print(df_feriados.head())
print(df_feriados.columns)
```

# Clima

```python
import requests
import pandas as pd

url = (
    "https://archive-api.open-meteo.com/v1/archive"
    "?latitude=-22.90"
    "&longitude=-43.20"
    "&start_date=2023-01-01"
    "&end_date=2023-01-03"
    "&hourly=temperature_2m,precipitation"
)

response = requests.get(url, timeout=30)
response.raise_for_status()

dados = response.json()

df_clima = pd.DataFrame({
    "data_hora": dados["hourly"]["time"],
    "temperatura": dados["hourly"]["temperature_2m"],
    "precipitacao": dados["hourly"]["precipitation"]
})

print(df_clima.head())
```
