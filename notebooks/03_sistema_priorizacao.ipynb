# Parte 3: Sistema de Priorização

## 9. Score de Prioridade

A proposta é utilizar no score o risco do modelo, a urgência do tipo de chamado, o impacto operacional/social, o clima e a equidade territorial.

Portanto, sobre o risco do modelo, será considerada a probabilidade de o chamado não ser resolvido em até 7 dias, tornando-se o principal componente do score, com peso de 40%.

A urgência do tipo de chamado recebe peso de 20%, uma vez que diferentes categorias possuem impactos distintos para a população. Chamados relacionados à infraestrutura urbana e serviços essenciais, como drenagem, iluminação pública e manejo arbóreo, tendem a exigir maior prioridade devido ao potencial impacto em segurança, mobilidade e qualidade de vida. Por isso, a definição de pesos específicos para cada tipo também foi considerada.

O impacto territorial, com peso de 15%, considera o volume histórico de chamados por região ou bairro, refletindo a pressão sobre os serviços públicos. Regiões com maior volume indicam maior demanda e, portanto, maior impacto coletivo caso os chamados não sejam tratados adequadamente.

A equidade territorial também recebe peso de 15%, com o objetivo de evitar que regiões mais complexas ou com menor previsibilidade operacional sejam negligenciadas. Esse componente permite balancear a priorização, garantindo que áreas com maior dificuldade estrutural ou histórica também sejam contempladas na alocação de recursos.

Por fim, o contexto climático, com peso de 10%, incorpora situações de condições adversas, como chuvas intensas ou temperaturas extremas, que podem agravar determinados tipos de chamados, especialmente aqueles relacionados à infraestrutura urbana.

Dessa forma, o score final combina eficiência e equidade, permitindo priorizar chamados com maior risco de atraso sem deixar de considerar o impacto social e territorial, direcionando a atuação do órgão público de forma mais estratégica e orientada por dados.

#### Resumo
Score Prioridade = 0,40 * risco_atraso + 0,20 * urgencia_tipo + 0,15 * impacto_territorial + 0,15 * equidade_territorial + 0,10 * contexto_climatico
